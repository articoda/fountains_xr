Create boundary file for Italy from the official European boundary file that we use to create the mask for the cerra data

```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
countries_file = r"c:\Users\ludog\Desktop\WORK\paper fontanelle\data\CNTR_RG_03M_2024_4326.gpkg"

countries = gpd.read_file(countries_file)
```

```python
print(countries.columns)

print(countries["CNTR_ID"].unique())
```

```python
italy = countries[countries["CNTR_ID"] == "IT"].copy()

print(italy)
```

```python
fig, ax = plt.subplots(figsize=(6,8))

italy.plot(
    ax=ax,
    edgecolor="black",
    facecolor="lightgreen"
)

plt.show()
```

```python
italy.to_file(r"c:\Users\ludog\Desktop\WORK\paper fontanelle\data\Italy_boundary.gpkg", driver="GPKG")
```
